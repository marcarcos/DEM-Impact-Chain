# Dataset exploration (`dataset.npz`)

Quick look at the random-design dataset written by `generate_dataset.py`
(360k nominal designs, equal thirds homogeneous / heterogeneous / heterogeneous-excluding-bistability,
design space of the paper: $c_2\in[-120,0]$, $c_3\in[0,4000]$, $c_3>\tfrac29 c_2^2$).

Keys: `c2`, `c3` (rows, 4) family coefficients A-D; `ratio` = KE ratio $KE_{non}/KE_{lin}$ (lower is better);
`category` 0/1/2 = homo/hetero/hetero_excl; `spring_type` (rows, 4) 0 monotonic, 1 snap-through, 2 bistable.

Cell 3 counts designs below the paper's Fig. 1d window edge (KE $\le$ 0.03) per category;
`good.py` turns these into the paper's robustness comparison.

```python
import pandas as pd
import numpy as np
import numba 
```

```python
d = np.load("dataset.npz")
X = np.hstack([d["c2"], d["c3"]])        # (360000, 8)
y = np.log(d["ratio"])                    # train on log: ratios span ~3 decades
cat, types = d["category"], d["spring_type"]
```

```python
good = d["ratio"] <= 0.03
for c, name in enumerate(["homo", "hetero", "hetero_excl"]):
    m = d["category"] == c
    print(f"{name:12s} {good[m].sum():5d} good  (best {d['ratio'][m].min():.4f})")
```
